```python
# DATOS DEL SOLICITANTE

# nombre completo
while True:
    nombre = input("Ingrese su nombre completo: ")
    if nombre.replace(" ", " ").isalpha and len(nombre.strip()) > 0:
        print(f"Nombre válido: {nombre}\n")
        break
    else:
        print("Error: El nombre solo puede contener letras y espacios. Inténtelo nuevamente.")

# Tipo de documento
while True:
    tipo_documento = input("Ingrese tipo de documento (CC, TI, CE, PAS): ").upper()
    if tipo_documento in ["CC", "TI", "CE", "PA"]:
        print(f"Tipo de documento: {tipo_documento}\n")
        break
    else:
        print("Error: Tipo de documento no válido. Use: CC, TI, CE o PA.")

# Número de documento
while True:
    numero_documento = input("Ingrese número de documento: ")
    if numero_documento.isdigit() and len(numero_documento) >= 5:
        print(f"Número de documento válido: {numero_documento}\n")
        break
    else:
        print("Error: El documento solo puede contener números y debe tener al menos 5 dígitos.")

# tipo de telefono
while True:
    tipo_tel = input("¿Tipo de teléfono? (CEL / FIJO): ").upper()
    if tipo_tel in ["CEL", "FIJO"]:
        print(f"Tipo de teléfono: {tipo_tel}\n")
        break
    else:
        print("Error: Escriba 'CEL' o 'FIJO'.")
# numero de telefono
while True:
    telefono = input("Ingrese número de teléfono: ")
    if telefono.isdigit() and len(telefono) >= 7:
        print(f"Teléfono válido: {telefono}\n")
        break
    else:
        print("Error: El teléfono solo puede contener números y debe tener al menos 7 dígitos.")

# correo electronico
while True:
    correo = input("Ingrese su correo electrónico: ")
    if "@" in correo and "." in correo and correo.index("@") < correo.rfind("."):
        print(f"Correo válido: {correo}\n")
        break
    else:
        print("Error: Ingrese un correo válido (debe contener @ y .).")

# dirección
while True:
    direccion = input("Ingrese su dirección de residencia: ")
    if len(direccion.strip()) > 0:
        print(f"Dirección registrada: {direccion}\n")
        break
    else:
        print("Error: La dirección no puede estar vacía.")


# DATOS INGRESADOS

print("DATOS DEL SOLICITANTE")
print("-" * 50)

print(f"Nombre completo:     {nombre}")
print(f"Documento:           {tipo_documento} - {numero_documento}")
print(f"Teléfono:            {tipo_tel} - {telefono}")
print(f"Correo electrónico:  {correo}")
print(f"Dirección:           {direccion}")
print("Todos los datos han sido registrados correctamente.")
```

```python
# INFORMACION DE LA PQRS
# Tipo de solicitud
while True:
    seleccionar = input("Ingrese el tipo de solicitud (1.Peticion, 2.Queja, 3.Reclamo, 4.Sugerencia): ")
    if seleccionar == "1":
        tipo_solicitud = "Peticion"
        break
    elif seleccionar == "2":
        tipo_solicitud = "Queja"
        break
    elif seleccionar == "3":
        tipo_solicitud = "Reclamo"
        break
    elif seleccionar == "4":
        tipo_solicitud = "Sugerencia"
        break
    else:
        print("Numero invalido(Intentelo de nuevo) ")

#Fecha de radicacion
import datetime as dt

def pedir_fecha():
    while True:
        dia = input("Ingrese fecha de radicacion (DD/MM/AAAA): ")
        try:
            fecha_de_radicacion = dt.datetime.strptime(dia, "%d/%m/%Y")
            hoy = dt.datetime.now()
            if fecha_de_radicacion > hoy:
                print("La fecha debe ser de hoy o anterior: ")
            else:
                return fecha_de_radicacion
        except ValueError:
            print("Formato de fecha incorrecto. Inténtelo de nuevo.")

fecha_registro = pedir_fecha()
fecha_respuesta = fecha_registro + dt.timedelta(days=30)

print("=" * 50)
print("Informacion de la PQRS: ")
print("=" * 50)
print(f"Tipo de solicitud: {tipo_solicitud}")
print(f"Fecha de radicacion: {fecha_registro.strftime('%d/%m/%Y')}")
print(f"Fecha de respuesta: {fecha_respuesta.strftime('%d/%m/%Y')}")
print("=" * 50)
```

```python
# CAPTURA Y REPORTES

from collections import Counter
from datetime import date, datetime, timedelta

DIAS_PROXIMOS_A_VENCER = 5
ESTADO_SOLUCIONADA = "solucionada"

def obtener_dato(registro, campo, valor_predeterminado=None):
    """Lee un campo desde un diccionario o desde un objeto PQRS."""
    if isinstance(registro, dict):
        return registro.get(campo, valor_predeterminado)
    return getattr(registro, campo, valor_predeterminado)

def convertir_fecha(valor):
    """Convierte fechas date, datetime o texto DD/MM/AAAA o AAAA-MM-DD."""
    if isinstance(valor, datetime):
        return valor.date()

    if isinstance(valor, date):
        return valor

    if isinstance(valor, str):
        for formato in ("%d/%m/%Y", "%Y-%m-%d"):
            try:
                return datetime.strptime(valor.strip(), formato).date()
            except ValueError:
                continue
    return None

def calcular_promedio_dias_respuesta(registros):
    """Promedia los dias transcurridos en las PQRS solucionadas."""
    duraciones = []

    for registro in registros:
        estado = obtener_dato(registro, "estado_peticion", "")
        if not isinstance(estado, str) or estado.casefold() != ESTADO_SOLUCIONADA.casefold():
            continue

        fecha_inicio = convertir_fecha(
            obtener_dato(registro, "fecha_radicacion")
        )
        fecha_solucion = convertir_fecha(
            obtener_dato(registro, "fecha_solucion")
        )

        if fecha_inicio and fecha_solucion and fecha_solucion >= fecha_inicio:
            duraciones.append((fecha_solucion - fecha_inicio).days)

    if not duraciones:
        return None

    return sum(duraciones) / len(duraciones)


def contar_por_campo(registros, campo):
    """Cuenta cuantas PQRS hay por cada valor de un campo."""
    conteo = Counter()

    for registro in registros:
        valor = obtener_dato(registro, campo)
        if valor:
            conteo[valor] += 1

    return dict(sorted(conteo.items()))


def obtener_pqrs_mas_antigua_pendiente(registros):
    """Busca la PQRS no solucionada con la fecha de radicación más antigua."""
    pendientes = []

    for registro in registros:
        estado = obtener_dato(registro, "estado_peticion", "Registrada")
        fecha = convertir_fecha(
            obtener_dato(registro, "fecha_radicacion")
        )

        if (
            isinstance(estado, str)
            and estado.casefold() != ESTADO_SOLUCIONADA.casefold()
            and fecha is not None
        ):
            pendientes.append((fecha, registro))

    if not pendientes:
        return None

    return min(pendientes, key=lambda elemento: elemento[0])[1]

def obtener_pqrs_proximas_a_vencer(registros, hoy=None):
    """Lista las PQRS pendientes que vencen hoy o en los próximos cinco días."""
    fecha_actual = hoy or date.today()
    fecha_limite = fecha_actual + timedelta(days=DIAS_PROXIMOS_A_VENCER)
    proximas = []

    for registro in registros:
        estado = obtener_dato(registro, "estado_peticion", "Registrada")
        fecha_maxima = convertir_fecha(
            obtener_dato(registro, "fecha_max_respuesta")
        )

        if (
            isinstance(estado, str)
            and estado.casefold() != ESTADO_SOLUCIONADA.casefold()
            and fecha_maxima is not None
            and fecha_actual <= fecha_maxima <= fecha_limite
        ):
            proximas.append(registro)

    return proximas


def generar_estadisticas(registros, hoy=None):
    """Reúne el promedio y cinco estadísticas de las PQRS."""
    registros = list(registros)

    return {
        "promedio_dias_respuesta": calcular_promedio_dias_respuesta(registros),
        "cantidad_por_tipo_solicitud": contar_por_campo(
            registros, "tipo_solicitud"
        ),
        "cantidad_por_tipo_mascota": contar_por_campo(
            registros, "tipo_mascota"
        ),
        "pqrs_mas_antigua_pendiente": obtener_pqrs_mas_antigua_pendiente(
            registros
        ),
        "pqrs_proximas_a_vencer": obtener_pqrs_proximas_a_vencer(
            registros, hoy
        ),
        "cantidad_por_campus": contar_por_campo(
            registros, "campus_relacionado"
        ),
    }
```
